```python
import time
import math

# 1. ฟังก์ชัน Binary Search (O(log n))
# ค้นหาตำแหน่งของ target ใน list ที่เรียงลำดับแล้ว
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    steps = 0  # เพิ่มตัวนับรอบเพื่อดูว่าทำงานไปกี่ครั้งจริง ๆ
    
    while low <= high:
        steps += 1
        mid = (low + high) // 2  # หาตำแหน่งตรงกลาง (แบ่งครึ่งปัญหา)
        guess = arr[mid]
        
        if guess == target:
            return mid, steps
        if guess > target:
            high = mid - 1       # ตัดครึ่งขวาทิ้งไป
        else:
            low = mid + 1        # ตัดครึ่งซ้ายทิ้งไป
            
    return -1, steps

# =======================================================
# ทดสอบระบบจับเวลาจริงในแต่ละขนาดข้อมูล
# =======================================================

# เคสที่ 1: ข้อมูลขนาดเล็ก (10 ตัว) -> ค้นหาเลขตัวสุดท้าย
small_list = list(range(10))
target_small = 9

start_small = time.perf_counter()
idx_small, steps_small = binary_search(small_list, target_small)
end_small = time.perf_counter()
time_small = end_small - start_small

# เคสที่ 2: ข้อมูลขนาดใหญ่ (1,000,000 ตัว) -> ค้นหาเลขตัวสุดท้าย
large_list = list(range(1000000))
target_large = 999999

start_large = time.perf_counter()
idx_large, steps_large = binary_search(large_list, target_large)
end_large = time.perf_counter()
time_large = end_large - start_large

# =======================================================
# แสดงผลลัพธ์
# =======================================================
print(f"Small List (10 ตัว)    ---> วิ่งไป: {steps_small} รอบ | เวลาที่ใช้: {time_small:.8f} วินาที")
print(f"Large List (1 ล้านตัว)  ---> วิ่งไป: {steps_large} รอบ | เวลาที่ใช้: {time_large:.8f} วินาที")
```

## 📌 1. สรุปเปรียบเทียบเชิงทฤษฎี

* **\(O(1)\) - Constant Time:** ดีที่สุดในอุดมคติ เวลาคงที่เสมอไม่ว่าข้อมูลจะมากแค่ไหน เพราะเข้าถึงตำแหน่งข้อมูลโดยตรง
* **\(O(\log n)\) - Logarithmic Time:** ดีเยี่ยมเป็นอันดับสอง เวลาเพิ่มขึ้นช้ามากเมื่อข้อมูลโตขึ้น เพราะขนาดของปัญหาถูกตัดลดลงครึ่งหนึ่งในทุกรอบการทำงาน
* **\(O(n)\) - Linear Time:** ประสิทธิภาพปานกลาง เวลาเพิ่มขึ้นเป็นเส้นตรงตามขนาดข้อมูล ยิ่งข้อมูลเยอะ ลูปยิ่งต้องทำงานมากตามตัวแปร \(n\)

---

## 📊 2. เปรียบเทียบประสิทธิภาพให้เห็นภาพชัดเจน

หากเรามีข้อมูลอินพุตจำนวน **1,000,000 รายการ** (\(n = 1,000,000\)) และต้องการค้นหาข้อมูลตัวที่อยู่ท้ายสุด:
* **โค้ดแบบทั่วไป (\(O(n)\)):** ตัวประมวลผลอาจต้องวิ่งไล่ทำงานวนลูปตรวจสอบยาวนานถึง **1,000,000 รอบ**
* **โค้ดแบบแบ่งครึ่งปัญหา (\(O(\log n)\) ฐาน 2):** จะใช้การประมวลผลสูงสุดเพียงแค่ประมาณ **20 รอบ** เท่านั้น!

---

## 💡 3. ตัวอย่างเหตุการณ์ในชีวิตประจำวัน (Real-world Analogy)

* **แบบ \(O(1)\):** เพื่อนส่งหนังสือมาให้คุณ 1 เล่ม แล้วถามว่า *"หน้าแรกสุดคือเลขหน้าอะไร"* คุณเปิดดูหน้าแรกได้ทันทีใน 1 วินาที ไม่ว่าหนังสือเล่มนั้นจะมี 10 หน้า หรือ 1,000,000 หน้าก็ตาม
* **แบบ \(O(\log n)\):** คุณเปิดพจนานุกรมเล่มหนาเพื่อหาคำว่า "Software" คุณเปิดสุ่มไปตรงกลางเล่ม ถ้าเจอหมวดตัว M คุณรู้ทันทีว่าตัว S ต้องอยู่ครึ่งเล่มขวา จึงตัดครึ่งเล่มซ้ายทิ้งไป แล้วเปิดสุ่มตรงกลางของครึ่งเล่มขวาต่อ ทำซ้ำแบบนี้ไปเรื่อยๆ จนเจอคำที่ต้องการได้อย่างรวดเร็ว
* **แบบ \(O(n)\):** คุณต้องเปิดหาคำเฉพาะคำหนึ่งในหนังสือ โดยคุณไม่รู้ว่าอยู่หน้าไหนและไม่มีดัชนีบอก คุณจึงต้องจำใจเปิดอ่านทีละหน้าตั้งแต่หน้าแรกไปเรื่อยๆ จนกว่าจะเจอคำนั้น

## 💻 4. ตัวอย่างการทำงานในระบบคอมพิวเตอร์: การหารากที่สอง (Square Root)

ในระบบคอมพิวเตอร์ การหาค่ารากที่สองของตัวเลขขนาดยักษ์ (เช่น \(\sqrt{987,654,321}\)) อัลกอริทึมที่มีประสิทธิภาพสูงจะไม่ใช้การสุ่มเดาเพิ่มค่าขึ้นทีละ 1 แบบเส้นตรง (\(O(n)\)) แต่จะใช้วิธี **Binary Search Style Square Root** โดยการสุ่มค่าตรงกลางในช่วงคำตอบที่เป็นไปได้ แล้วนำมาคำนวณดูว่าเกินค่าที่ต้องการหรือไม่ เพื่อ "ตัดขอบเขตคำตอบที่ไม่ใช่ออกไปทีละครึ่งหนึ่ง" ในทุกขั้นตอนการคิด ทำให้คอมพิวเตอร์ได้คำตอบทศนิยมที่แม่นยำออกมาในเวลาไม่กี่รอบการประมวลผล

## 🧑‍💻 5. บล็อกโค้ดพล็อตกราฟเปรียบเทียบ (Python)
คัดลอกโค้ดสคริปต์นี้ไปวางในบล็อกโค้ดบน Google Colab เพื่อวาดกราฟวิเคราะห์ความเร็วเชิงเปรียบเทียบ

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. จำลองขนาดข้อมูล (n) ตั้งแต่ 1 ถึง 100 รายการ เพื่อให้เห็นส่วนโค้งของกราฟได้ชัดเจน
n = np.arange(1, 101)

# 2. คำนวณค่าของแต่ละ Big Oตามสมการคณิตศาสตร์
y_o1 = np.ones(len(n))       # O(1) คงที่เท่ากับ 1 เสมอ
y_ologn = np.log2(n)        # O(log n) ฐาน 2
y_on = n                    # O(n) เพิ่มเป็นเส้นตรงตาม n

# 3. เริ่มพล็อตกราฟด้วย Matplotlib
plt.figure(figsize=(12, 7))

plt.plot(n, y_on, label='O(n) - Linear Time', color='#d62728', linewidth=2.5)
plt.plot(n, y_ologn, label='O(log n) - Logarithmic Time', color='#ff7f0e', linewidth=2.5)
plt.plot(n, y_o1, label='O(1) - Constant Time', color='#1f77b4', linewidth=2.5)

# 4. ตกแต่งหน้าตากราฟให้เป็นสากลและเข้าใจง่าย
plt.title('Big O Complexity Comparison: O(1) vs O(log n) vs O(n)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Input Size (n)', fontsize=12)
plt.ylabel('Operations / Time Steps', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=12)

# บังคับให้แกน Y เริ่มจาก 0
plt.ylim(0, 105)

# แสดงกราฟออกมาบนหน้าจอ Colab
plt.show()
```

## 🔍 สิ่งที่ควรสังเกตเมื่อกราฟแสดงผล
* **เส้นสีแดง $O(n)$:** จะพุ่งทะยานชันขึ้นเป็นเส้นตรง 45 องศาอย่างรวดเร็ว สะท้อนภาระงานที่เพิ่มขึ้นเป็นเงาตามตัว
* **เส้นสีส้ม $O(\log n)$:** ในช่วงแรกจะชันขึ้นเล็กน้อย แต่หลังจากนั้นจะค่อยๆ ลาดลงและนอนแนวราบลงเรื่อยๆ แม้ $n$ จะเพิ่มขึ้นเท่าไร เส้นนี้ก็แทบจะไม่วิ่งขึ้นข้างบนเลย
* **เส้นสีน้ำเงิน $O(1)$:** นอนราบนิ่งติดขอบล่างสุดขนานไปกับพื้น ไม่สะทกสะท้านต่อค่า $n$ ที่เพิ่มขึ้น
