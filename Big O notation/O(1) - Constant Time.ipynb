# 🚀 **ประเภทของ Big O** *(เรียงจากดีที่สุด ตามลำดับ)*

### 1. O(1) - Constant Time
เป็น **Big O ที่ดีที่สุด** ไม่ว่า input จะมีขนาดเท่าไร ระยะเวลาในการประมวลผลจะคงที่เสมอ

#### 💡 ตัวอย่างรูปแบบการทำงาน:
* การเข้าถึงข้อมูลใน array ด้วย index
* การคำนวณทางคณิตศาสตร์พื้นฐาน (arithmetic operations: `+`, `-`, `*`, `/`)
* การค้นหาข้อมูลด้วย Key ใน Hash Table (Dictionary / Map)


```python
import time

def get_first_element(my_list):
    # เข้าถึงตำแหน่งที่ 0 ทันที ไม่ว่าลิสต์จะยาวแค่ไหน (O(1))
    return my_list[0]

# 1. เตรียมข้อมูลที่มีขนาดต่างกันมหาศาล
small_list = [10, 20, 30]
large_list = list(range(1000000))  # ลิสต์ที่มีข้อมูล 1 ล้านตัว

# 2. วัดเวลาของ Small List
start_time = time.perf_counter()  # เริ่มจับเวลา (หน่วยเป็นวินาทีที่มีความแม่นยำสูง)
result_small = get_first_element(small_list)
end_time = time.perf_counter()    # สิ้นสุดการจับเวลา
time_small = end_time - start_time

# 3. วัดเวลาของ Large List
start_time = time.perf_counter()  # เริ่มจับเวลาใหม่
result_large = get_first_element(large_list)
end_time = time.perf_counter()    # สิ้นสุดการจับเวลา
time_large = end_time - start_time

# 4. แสดงผลลัพธ์และเวลาที่ใช้
print(f"Small List (3 ตัว) -> ผลลัพธ์: {result_small} | เวลาที่ใช้: {time_small:.8f} วินาที")
print(f"Large List (1 ล้านตัว) -> ผลลัพธ์: {result_large} | เวลาที่ใช้: {time_large:.8f} วินาที")
```

# ⏳ O(1) - Constant Time Explanation

ความหมายของมันคือ **ไม่ว่าข้อมูลที่คุณใส่เข้ามา (Input Size หรือ $n$) จะมีขนาดใหญ่ขึ้นเป็นแสน ล้าน หรือพันล้านรายการ** ระยะเวลาและทรัพยากรที่คอมพิวเตอร์ใช้ในการประมวลผลคำสั่งนั้นจะ **"คงที่" (Constant) เสมอ** โดยไม่ผันแปรไปตามขนาดของข้อมูลครับ


# 📊 การทำงานของ O(1) ในรูปแบบกราฟ

### เปรียบเทียบกราฟความเร็ว (Time vs Input Size)

หากเรานำข้อมูลมาเขียนกราฟ โดยมีข้อกำหนดดังนี้:
* **แกน X:** ขนาดของข้อมูล ($n$)
* **แกน Y:** เวลาที่ใช้ประมวลผล (Time)

เส้นกราฟของ **$O(1)$ จะเป็นเส้นตรงแนวนอนขนานไปกับแกน X เสมอ** เพราะขนาดของ $n$ ไม่มีผลต่อความสูงของแกน Y เลย


# 💡 ตัวอย่างเหตุการณ์ในชีวิตประจำวัน (Real-world Analogy)

* **แบบ $O(1)$:**
  คุณมีหนังสือ 1 เล่ม (หรือ 1 ล้านเล่ม) อยู่ในมือ แล้วมีคนถามว่า *"หน้าแรกของหนังสือเล่มนี้คือหน้าอะไร?"* คุณเปิดไปหน้าแรกทันทีโดยใช้เวลาแค่ 1 วินาทีเท่ากัน ไม่ว่าหนังสือจะหนาแค่ไหน

* **แบบที่ไม่ใช่ $O(1)$:**
  คุณต้องเปิดหาคำเฉพาะคำหนึ่งในหนังสือที่มี 1 ล้านหน้า โดยต้องเปิดอ่านทีละหน้าไปเรื่อยๆ จนกว่าจะเจอ แบบนี้ยิ่งหนังสือเล่มใหญ่ขึ้น คุณก็ยิ่งใช้เวลานานขึ้น ($O(n)$)


# 💻 ตัวอย่างการทำงานในระบบคอมพิวเตอร์

* **การเข้าถึงข้อมูลใน Array ด้วย Index (`array[5]`):**
  คอมพิวเตอร์ไม่ได้วิ่งไล่หาตั้งแต่ตัวแรก แต่จะใช้วิธีคำนวณคณิตศาสตร์เบื้องหลังเพื่อกระโดดไปยังที่อยู่ของข้อมูล (Memory Address) ตัวนั้นโดยตรงทันที
  
* **การคำนวณทางคณิตศาสตร์พื้นฐาน (`+`, `-`, `*`, `/`):**
  ไม่ว่าคุณจะสั่ง `1 + 1` หรือ `987,654 + 321,098` CPU ของคอมพิวเตอร์จะใช้รอบสัญญาณนาฬิกา (Clock Cycles) ในจำนวนที่จำกัดและคงที่ในการประมวลผลผลลัพธ์ออกมาระดับฮาร์ดแวร์
  
* **การเพิ่ม/ลบข้อมูลที่ท้ายลิสต์ (Push/Pop ใน Stack):**
  การใส่ข้อมูลตัวใหม่ต่อท้ายแถวที่มีการล็อกตำแหน่งไว้แล้ว หรือหยิบตัวบนสุดออก สามารถทำได้ทันทีโดยไม่ต้องไปยุ่งหรือขยับตำแหน่งข้อมูลตัวอื่นๆ ในระบบเลย

### 📝 สรุป
การออกแบบโค้ดให้ทำงานได้ในระดับ **$O(1)$ เป็นเป้าหมายสูงสุดของนักพัฒนา** เพราะมันการันตีว่าระบบจะทำงานได้เร็วเท่าเดิมเสมอ แม้ว่าระบบจะมีผู้ใช้งานหรือข้อมูลเพิ่มขึ้นมหาศาลในอนาคต


```python
import time
# คลังข้อมูลพนักงาน (Dictionary)
employee_db = {
    "EMP001": "Alice",
    "EMP002": "Bob",
    "EMP999": "Zack"
    # สมมติต่อให้มีพนักงานเพิ่มขึ้นเป็น 10,000,000 คน...
}

def find_employee_name(emp_id):
    # ค้นหาด้วย Key ตรงๆ เป็น O(1) เพราะคอมพิวเตอร์แปลง Key เป็นที่อยู่หน่วยความจำทันที
    return employee_db.get(emp_id, "Not Found")

# 2. วัดเวลาของ Small List
start_time = time.perf_counter()  # เริ่มจับเวลา (หน่วยเป็นวินาทีที่มีความแม่นยำสูง)
result_small = get_first_element(small_list)
end_time = time.perf_counter()    # สิ้นสุดการจับเวลา
time_small = end_time - start_time

# 3. วัดเวลาของ Large List
start_time = time.perf_counter()  # เริ่มจับเวลาใหม่
result_large = get_first_element(large_list)
end_time = time.perf_counter()    # สิ้นสุดการจับเวลา
time_large = end_time - start_time

# ค้นหาเจอทันทีโดยไม่ต้องไล่ดูทีละคนตั้งแต่ต้นจนจบ
print(find_employee_name("EMP002")) # ผลลัพธ์: Bob
```

การหาข้อมูลด้วยคีย์ใน Hash Table (Dictionary / Map)หนึ่งในโครงสร้างข้อมูลที่เป็น \(O(1)\) ที่ถูกใช้บ่อยที่สุดในการเขียนโปรแกรมเพื่อหลีกเลี่ยงการใช้ Loop ค้นหาข้อมูล

💡 ข้อสังเกต: ทำไมถึงเป็น \(O(1)\)?จุดร่วมของโค้ดทุกตัวอย่างข้างต้นคือ "ไม่มีการใช้ Loop (เช่น for หรือ while)" ที่แปรผันตามขนาดของข้อมูลเลย โค้ดจะทำงานบรรทัดเดิม ขั้นตอนเท่าเดิมเสมอเมื่อกดรันครับคุณต้องการให้ลองเปรียบเทียบโค้ดแบบ \(O(1)\) นี้ กับโค้ดแบบ \(O(n)\) (Linear Time) ที่ต้องใช้ Loop ไล่หาข้อมูล เพื่อให้เห็นความแตกต่างของความเร็วเมื่อข้อมูลมีขนาดใหญ่ขึ้น และกรณีมีโครงสร้างข้อมูลตัวไหนที่อยากเจาะลึกเป็นพิเศษ

## 📌 สรุปเปรียบเทียบเชิงทฤษฎี
* **$O(1)$ - Constant Time:** ประสิทธิภาพคงที่เสมอ ไม่ขึ้นอยู่กับขนาดข้อมูล เพราะใช้วิธีคำนวณ Address ตรงๆ ไม่มีลูปขยายตัว
* **$O(n)$ - Linear Time:** ประสิทธิภาพลดลงเป็นเส้นตรงตามขนาดข้อมูล ยิ่งอินพุต ($n$) เยอะ ลูป (`for`/`while`) ก็ต้องทำงานมากขึ้นเป็นเงาตามตัว

---

## 🧑‍💻 บล็อกโค้ดทดลองจับเวลาและพล็อตกราฟเปรียบเทียบ (Python)


```python
import time

# ---------------------------------------------------------
# 1. ฟังก์ชันแบบ O(1) - ดึงข้อมูลด้วย Index โดยตรง
# ---------------------------------------------------------
def find_last_element_o1(my_list):
    return my_list[-1]  # กระโดดไปตัวสุดท้ายทันที

# ---------------------------------------------------------
# 2. ฟังก์ชันแบบ O(n) - ใช้ Loop วิ่งไล่หาจนกว่าจะเจอตัวสุดท้าย
# ---------------------------------------------------------
def find_last_element_on(my_list, target):
    for item in my_list:
        if item == target:
            return item
    return None

# =========================================================
# เริ่มการทดลองด้วยข้อมูลขนาดใหญ่ (10,000,000 ตัว)
# =========================================================
large_list = list(range(10000000))
target_value = 9999999  # ข้อมูลตัวสุดท้ายในลิสต์

# ทดสอบ O(1)
start_o1 = time.perf_counter()
res_o1 = find_last_element_o1(large_list)
end_o1 = time.perf_counter()
time_o1 = end_o1 - start_o1

# ทดสอบ O(n)
start_on = time.perf_counter()
res_on = find_last_element_on(large_list, target_value)
end_on = time.perf_counter()
time_on = end_on - start_on

# =========================================================
# แสดงผลลัพธ์
# =========================================================
print(f"🎯 ผลลัพธ์แบบ O(1) -> เจอ: {res_o1} | เวลาที่ใช้: {time_o1:.8f} วินาที")
print(f"🎯 ผลลัพธ์แบบ O(n) -> เจอ: {res_on} | เวลาที่ใช้: {time_on:.8f} วินาที")
print(f"⚡ โค้ดแบบ O(1) ทำงานเร็วกว่า O(n) ประมาณ {int(time_on / max(time_o1, 1e-9)):,} เท่า!")
```

```python
import time
import matplotlib.pyplot as plt
import numpy as np

# 1. นิยามฟังก์ชัน O(1) (เข้าถึงตัวสุดท้ายผ่าน Index ทันที)
def find_last_o1(my_list):
    return my_list[-1]

# 2. นิยามฟังก์ชัน O(n) (ใช้ลูปวิ่งไล่เช็คตั้งแต่ตัวแรกจนถึงตัวสุดท้าย)
def find_last_on(my_list, target):
    for item in my_list:
        if item == target:
            return item
    return None

# 3. กำหนดขนาดอินพุต (n) ที่เพิ่มขึ้นเรื่อยๆ จากหลักร้อยจนถึงหลักล้าน
input_sizes = [100, 1000, 10000, 50000, 100000, 500000, 1000000]
times_o1 = []
times_on = []

print("⏳ เริ่มต้นการทดลองประมวลผลจริง...")

for size in input_sizes:
    # สร้างลิสต์ข้อมูลทดสอบขนาด n
    test_list = list(range(size))
    target_value = size - 1  # ข้อมูลตัวสุดท้ายที่ต้องการหา

    # ทดสอบและจับเวลาฟังก์ชัน O(1) [หน่วย: นาโนวินาที]
    start_o1 = time.perf_counter_ns()
    find_last_o1(test_list)
    end_o1 = time.perf_counter_ns()
    times_o1.append((end_o1 - start_o1) / 1000) # แปลงเป็น Microseconds

    # ทดสอบและจับเวลาฟังก์ชัน O(n) [หน่วย: นาโนวินาที]
    start_on = time.perf_counter_ns()
    find_last_on(test_list, target_value)
    end_on = time.perf_counter_ns()
    times_on.append((end_on - start_on) / 1000) # แปลงเป็น Microseconds

    print(f"ขนาดอินพุต (n): {size:10,} รายการ | O(1) ใช้: {times_o1[-1]:.3f} μs | O(n) ใช้: {times_on[-1]:.3f} μs")

print("✅ การทดลองเสร็จสิ้น! กำลังวาดกราฟสถิติ...")

# 4. พล็อตกราฟเปรียบเทียบผลลัพธ์ด้วย Matplotlib
plt.figure(figsize=(12, 6))

# พล็อตเส้น O(1) และ O(n)
plt.plot(input_sizes, times_on, marker='x', color='#d62728', linewidth=2.5, label='O(n) - Linear Loop Search')
plt.plot(input_sizes, times_o1, marker='o', color='#1f77b4', linewidth=2.5, label='O(1) - Constant Index Access')

# ปรับแต่งหน้าตากราฟ
plt.title('Performance Showdown: O(1) Constant Time vs O(n) Linear Time', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Input Size (n)', fontsize=12)
plt.ylabel('Execution Time (Microseconds)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=12)

# แสดงกราฟบน Google Colab
plt.show()
```

1. **เส้นสีน้ำเงิน $O(1)$:** จะนอนราบนิ่งสนิทขนานไปกับพื้นด้านล่าง บ่งบอกว่าต่อให้ข้อมูลจะทะลุไป 1 ล้านตัว เวลาที่ใช้ก็ยังน้อยกว่า 1 ไมโครวินาทีเท่าเดิม
